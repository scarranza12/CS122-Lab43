```python
N = 8

for i in range(2, N):  # +1 would include N total number
    if N % i == 0:
        print(f'{N} is not a prime number.')
        break
print(i)

if i == N - 1: ## MEANING: checking to see for loop has full-iterations
        print(f'{N} is a prime number.')
        
# ( i ) is the value used in for loop
```
